This notebook demonstrates core Python concepts needed for data engineering.

```python
# 01 Variables and Data Types
batch_size = 1000
data_source = "s3://bucket/data/"
is_production = False

print(f"Batch size: {batch_size}")
print(f"Data source: {data_source}")
```

```python
# 02 String Operations for Data Cleaning
def clean_column_name(col: str) -> str:
    """Clean column names for data pipelines."""
    return col.lower().strip().replace(" ", "_").replace("-", "_")
```

```python
# 03 List Comprehension for Filtering
customers = [
    {"name": "John", "age": 25, "status": "active"},
    {"name": "Jane", "age": 17, "status": "active"},
    {"name": "Bob", "age": 30, "status": "inactive"},
]

# Filter active customers aged 18+
active_adults = [c for c in customers if c["status"] == "active" and c["age"] >= 18]
print(f"Active adults: {active_adults}")
```

```python
# 04 Data Validation with Pydantic
from pydantic import BaseModel, EmailStr
from datetime import date

class Customer(BaseModel):
    customer_id: int
    first_name: str
    email: EmailStr
    signup_date: date
    age: int

customer = Customer(
    customer_id=1,
    first_name="John",
    email="john@example.com",
    signup_date="2024-01-01",
    age=30
)
print(customer.model_dump())
```
